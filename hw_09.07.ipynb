## 02 맷플롯립
### **맷플롯립의 pyplot 모듈의 이해**
- plot() : 입력값으로 x좌표에 해당하는 값과 y좌표에 해당하는 여러 데이터 값을 입력받아 x와 y값이 매핑되는 선 그래프 그리기
- title() : 입력값으로 타이틀 문자열 받아서 차트 타이틀 설정
- show() : 축을 포함한 선 그래프 화면에 나타낼 수 있음

```python
# pyplot 모듈을 (관용적으로) plt라는 이름으로 import
import matplotlib.pyplot as plt

# x좌표 입력값 [1, 2, 3], y좌표 입력값 [2, 4, 6]으로 선 그래프를 생성
plt.plot([1, 2, 3], [2, 4, 6]) 

# 선그래프 타이틀을 Hello plot으로 설정. 
plt.title("Hello plot") 

# 선 그래프를 출력
plt.show()
```

### **pyplot의 두가지 중요 요소 - Figure와 Axes 이해**
- Figure 객체 : 그림을 그리기 위한 캔버스 역할, 그림판 크기조절, 플롯 최종적으로 화면에 나타내는 역할 수행
- Axes : 실제 그림을 그리는 메서드, X축/Y축/타이틀/범례 등의 속성 설정
1) plt.plot([1,2,3],[2,4,6])
기본으로 설정된 Axes에서 Axes.plot() 함수를 호출하여 그림을 그림, Axes.plot()은 다시 line2D 객체를 호출하여 선 그래프 그림
2) plt.title("Hello plot")
내부적으로 다시 Axes.set_title() 함수를 호출하여 타이틀 설정

### **Figure와 Axis의 활용**
`Figure`
- 전체 그림판의 크기 조절하는데 사용
- plt모듈의 figure()함수를 호출하면 figure 객체가 반환
- 인자로 figsize=(가로크기, 세로크기) 입력하면 해당 크기를 가지는 Figure 객체를 반환
- 인자로 facecolor='색깔' 입력하면 배경색 변경 가능

`Axis`
- `axes()` : 현재 사용하는 Axes 객체 반환
- `plt.subplots()`: Figure와 Axes 객체를 함께 가져올 수 있음 / 인자로 여러개의 Axes 설정 가능
    - Figure 객체 하나와 Axes 객체 하나를 동시에 만들어서, 튜플로 묶어 반환하는 함수
      
      ex) fig, ax = plt.subplots() => 반환값을 fig, ax = ...처럼 두 변수에 나눠 받음
      
      cf) Figure(캔버스)만 있으면 그 위에 아무것도 그릴 수 없고, Axes(그림 그리는 영역)만 있으면 그게 어느 캔버스에 속하는지가 없음. 그래서 실제로 그래프를 그리려면 둘 다 필요한데, plt.subplots()는 이 둘을 한 번에 만들어서 짝지어 반환해주는 편의 함수

- `plot()` : 선그래프 그림
- `set_title` : 제목 설정

```python
# plt.figure()는 주로 figure의 크기를 조절하는 데 사용됨.
# 아래는 figure 크기가 가로 10, 세로 4인 Figure객체를 설정하고 반환함. 
plt.figure(figsize=(10, 4)) 

plt.plot([1, 2, 3], [2, 4, 6]) 
plt.title("Hello plot") 
plt.show()
```

```python
figure = plt.figure(figsize=(10, 4))
print(type(figure))
```

```python
plt.figure(figsize=(4, 4), facecolor='yellow')
plt.plot([1, 2, 3], [2, 4, 6]) 
plt.title("Hello plot") 
plt.show()
```

```python
ax = plt.axes()
print(type(ax))
```

```python
### pyplot에서 설정된 Figure와 Axes 객체를 함께 가져오기 
fig, ax = plt.subplots()
print('fig type:', type(fig), '\nax type:', type(ax))
```

```python
fig, ax = plt.subplots(figsize=(4, 4))
# Axes 객체의 plot()을 이용하여 선 그래프를 그림. 
ax.plot([1, 2, 3], [2, 4, 6])

# Axes 객체의 set_title( )을 이용하여 선 그래프 타이틀을 Hello plot으로 설정. 
ax.set_title('Hello plot')
plt.show()
```

### **여러 개의 plot을 가지는 subplot들을 생성하기**
`plt.subplots()`
- 주로 여러개의 subplot을 생성하는데 활용
- 하나의 Figure상에서 여러개의 그래프를 그릴 수 있게 만들어줌. 이때 개별 subplot은 하나의 Axes를 가짐
  
=> 여러개의 그래프를 그릴 수 있게하는 1개의 Figure와 여러 개의 Axes를 생성하고 반환

```python
# 아래는 두개의 subplot, 즉 두개의 Axes를 생성하되 열 방향으로 두개를 배치함. 
# 또한 전체 크기가 가로 6, 세로 3인 Figure 생성. 
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(6, 3))
```

### **pyplot의 plot() 함수를 이용해 선 그래프 그리기**
맷플롯립은 매우 많은 시각화 객체와 함수 제공

`plot()`
- pyplot의 plot() 함수는 선 그래프 그릴 때 활용
- X 좌표 값, Y 좌표 값으로 파이썬 리스트, 넘파이 ndarray, 판다스 DataFrame/Series 모두 적용 가능
- **X값과 Y값의 개수는 모두 같은 크기여야함**
- color 인자를 이용하여 선 색깔 변경 가능
- 그 외 인자 : marker/ linestyle / linewidth / markersize 

```python
import numpy as np
import pandas as pd

x_value = [1, 2, 3, 4]
y_value = [2, 4, 6, 8]
#x_value = np.array([1, 2, 3, 4]) # 넘파이 array로 테스트하려면 주석을 해제하고 실행
#y_value = np.array([2, 4, 6, 8]) # 넘파이 array로 테스트하려면 주석을 해제하고 실행

# 입력값으로 파이썬 리스트, numpy array 가능. x 좌표값과 y좌표값은 모두 같은 크기를 가져야 함. 
plt.plot(x_value, y_value)
plt.show()
```

```python
plt.plot(x_value, y_value, color='green')
```

```python
plt.plot(x_value, y_value, color='red', marker='o', linestyle='dashed', linewidth=2, markersize=12)
```

### **축 명칭 설정, 축의 눈금(틱)값 회전, 범례(lengend) 설정하기**
- xlabel(), ylabel() : 축의 명칭 설정
- xticks(), yticks() : X축/Y축 눈금값(틱값) 회전
- plt.legend() : 각 그래프에 붙여둔 label 값들을 모아서, 화면에 범례(legend) 상자로 표시
- plt.legend()를 빼면 초록 선, 빨간 선은 그대로 그려지지만 어떤 색이 뭘 의미하는지 알려주는 박스가 화면에 안 나타남
- "합친다"는 별도의 코드가 있는 게 아니라, plt.plot()을 두 번 호출하면 자동으로 같은 그래프 위에 쌓임

```python
plt.plot(x_value, y_value, color='red', marker='o', linestyle='dashed', linewidth=2, markersize=12)

# x축과 y축의 이름을 텍스트로 할당하여 설정합니다. 
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

```python
# x 좌표 값을 0 ~ 99까지 100개로 설정. 
x_value = np.arange(0, 100)
# y 좌표 값은 개별 x좌표값을 2배로 해서 100개로 설정. 
y_value = 2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

# x축 눈금값을 90도로 회전
plt.xticks(rotation=90)

plt.show()
```

```python
x_value = np.arange(0, 100)
y_value = 2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

# X축의 눈금값을 np.arange(0, 100, 5)로 하여 0 부터 99까지 5 스텝으로 20개를 표시하고 90도 회전  
plt.xticks(ticks=np.arange(0, 100, 5), rotation=90)

plt.title('Hello plot')
plt.show()
```

```python
# X좌표값은 0~99까지 100개, 첫번째 Y좌표값은 X좌표값의 두 배, 두번째 좌표값은 X좌표값의 네 배
x_value_01 = np.arange(0,100)
y_value_01 = 2*x_value_01
y_value_02 = 4*x_value_01

# X좌표값과 첫번째 Y좌표값으로 녹색의 선그래프를 그림. label을 'green line'으로 설정.
plt.plot(x_value_01, y_value_01, color='green', label='green line')
# X좌표값과 두번째 Y좌표값으로 빨간색 선그래프를 그림. label을 'red line'으로 설정.
plt.plot(x_value_01, y_value_02, color='red', label='red line')

plt.xlabel('x axis')
plt.ylabel('y axis')

# 개별 plot들의 label로 설정된 문자열을 범례로 표시
plt.legend()

plt.title('Hello plot')

plt.show()
```

```python
x_value_01 = np.arange(1, 10)
y_value_01 = 2*x_value_01

# 마커를 포함한 빨간선 대쉬 선 그래프를 그리고, label은 'red line'으로 설정  
plt.plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
# X값에 따른 Y값을 나타내는 초록색 막대 그래프를 그리고 label은 'bar plot'으로 설정
plt.bar(x_value_01, y_value_01, color='green', label='bar plot')
plt.xlabel('x axis')
plt.ylabel('y axis')
# 두개의 그래프에서 label설정된 문자열 값을 범례로 표시
plt.legend()

plt.title('Hello plot')

plt.show()
```

```python
# 복잡한 그래프를 그릴수록 객체지향 방식(ax)이 더 안전(plt보다)
figure = plt.figure()
# Axes 객체를 추출하여 ax변수에 할당. 
ax = plt.axes()

# plt.plot( )은 ax.plot( )으로 plt.bar( )는 ax.bar( )로 변경. 
ax.plot(x_value_01, y_value_01, color='red',  marker='o', linestyle='dashed', label='red line')
ax.bar(x_value_01, y_value_01, color='green', label='bar plot')

# plt.xlabel( )은 ax.set_xlabel( )로, plt.ylabel( )은 ax.set_ylabel( )로 변경
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')

# ax.set_legend()가 아니라 ax.legend()임. 
ax.legend()
# plt.title( )을 plt.set_title( )로 변경
ax.set_title('Hello plot')

plt.show()
```

### **여러개의 subplots들을 이용해 개별 그래프들을 subplot별로 시각화하기**
- subplots들을 이용하면 여러 그래프들을 한꺼번에 시각화 가능
- 

```python
# Axes 객체들을 배열 형태로 받아서 시각화
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(6, 3))
# 왼쪽은 ax[0] 오른쪽은 ax[1] 객체 가리킴
```

```python
print('ax type:', type(ax))
print('ax[0] type:', type(ax[0]))
print('ax[1] type:', type(ax[1]))
```

```python
import numpy as np

x_value_01 = np.arange(1, 10)
x_value_02 = np.arange(1, 20)
y_value_01 = 2 * x_value_01
y_value_02 = 2 * x_value_02

# figsize는 (6,3)인 Figure와 2개의 Axes를 가지는 subplots을 반환. 
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(6, 3))

# 개별 Axes객체들을 각각 이용하여 선 그래프와 막대그래프를 그림. 
ax[0].plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
ax[1].bar(x_value_02, y_value_02, color='green', label='bar plot')

# 개별 Axes객체 각각에 X축명을 설정. 
ax[0].set_xlabel('ax[0] x axis')
ax[1].set_xlabel('ax[1] x axis')

# 개별 Axes객체 각각에 범례를 설정 
ax[0].legend()
ax[1].legend()

# 개별 Axes객체 각각에 타이틀을 설정
ax[0].set_title('Hello line')
ax[1].set_title('Hello bar')

plt.show()
```

```python
import numpy as np

x_value_01 = np.arange(1, 10)
x_value_02 = np.arange(1, 20)
y_value_01 = 2 * x_value_01
y_value_02 = 2 * x_value_02

# figsize는 (6,6)인 Figure와 2x2개의 Axes를 가지는 subplots을 반환.
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(6, 6))

# 2x2 Axes객체들을 각각 이용하여 선 그래프와 막대그래프를 그림. 
ax[0,0].plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
ax[0,1].bar(x_value_02, y_value_02, color='green', label='green bar')
ax[1,0].plot(x_value_01, y_value_01, color='green', marker='o', linestyle='dashed', label='green line')
ax[1,1].bar(x_value_02, y_value_02, color='red', label='red bar')

# 개별 Axes객체 각각에 X축명을 설정. 
ax[0,0].set_xlabel('ax[0][0] x axis')
ax[0,1].set_xlabel('ax[0][1] x axis')
ax[1,0].set_xlabel('ax[1][0] x axis')
ax[1,1].set_xlabel('ax[1][1] x axis')

# 개별 Axes객체 각각에 범례를 설정 
ax[0,0].legend()
ax[0,1].legend() 
ax[1,0].legend()
ax[1,1].legend() 

plt.show()
```

```python
import numpy as np

x_value_01 = np.arange(1, 10)
x_value_02 = np.arange(1, 20)
y_value_01 = 2 * x_value_01
y_value_02 = 2 * x_value_02

# figsize는 (6,6)인 Figure와 2x2개의 Axes를 가지는 subplots을 반환.
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(6, 6))

# 2x2 Axes객체들을 각각 이용하여 선 그래프와 막대그래프를 그림. 
ax[0][0].plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
ax[0][1].bar(x_value_02, y_value_02, color='green', label='green bar')
ax[1][0].plot(x_value_01, y_value_01, color='green', marker='o', linestyle='dashed', label='green line')
ax[1][1].bar(x_value_02, y_value_02, color='red', label='red bar')

# 개별 Axes객체 각각에 X축명을 설정. 
ax[0][0].set_xlabel('ax[0][0] x axis')
ax[0][1].set_xlabel('ax[0][1] x axis')
ax[1][0].set_xlabel('ax[1][0] x axis')
ax[1][1].set_xlabel('ax[1][1] x axis')

# 개별 Axes객체 각각에 범례를 설정 
ax[0][0].legend()
ax[0][1].legend() 
ax[1][0].legend()
ax[1][1].legend() 

plt.show()
```
